Show how to use a iterator. First, use enumerate to create a new tuple from above list

```
#use of a iterator
#iterator- returns numbers that starts with 25 and each and every time it is passed it gets squared.

class MyInit:    #for initialization
  def __init__(self):
        self.a = 25

  def __iter__(self):
    return self

  def __next__(self):  # does the operation
    x = self.a
    self.a = self.a * self.a
    return x

obj1 = MyInit()
itr = iter(obj1)

print(next(itr))
print(next(itr))
print(next(itr))
```

```
#First, use enumerate to create a new tuple from above list
listSequence = ["eat","sleep","repeat"]

tup = tuple(enumerate(listSequence))

#at first Enumerating thro' the list and then iterating thro' the loop with for loop.

print(tup)
```

Now use a zip iterator to do the same.

```
listSequence = ["eat", "sleep", "repeat"]

tup = tuple(zip(range(len(listSequence)), listSequence))

print(tup)
```

Now state the differences between a zip and enumerate iterator.

Differences Between Enumerate and Zip Iterator

Enumerate Iterator

1. Used to iterate over a single iterable while keeping track of its index.
2. Takes only one iterable as input.
3. Returns values in the form (index, value).
4. Automatically generates indices for the elements.
5. Useful when both the position and value of an element are needed.

Zip Iterator

1. Used to combine elements from two or more iterables.
2. Takes multiple iterables as input.
3. Returns tuples containing corresponding elements from the iterables.
4. Does not generate indices automatically.
5. Useful when multiple sequences need to be processed together.


Python`enumerate()` and `zip()` are built-in functions that can be used on sequences.

`enumerate()` function can be use to iterate over indices and items of a list.

`zip()` function make an iterator that aggregates elements from each of the iterables. In simple words it can be used to iterate over two or more lists in parallel.

# Working with CSV

Let's import a file with fuel economy data

```
import csv

with open('/content/mpg.csv', 'r') as csvfile:
    mpg = list(csv.DictReader(csvfile))
```

Find out how many items are there in the dictionary  

```
count = 0

for item in mpg[1].items():
    count += 1

print(count)
```

Find the column names with keys

```
print(mpg[0].keys())
```

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

```
total = 0

for car in mpg:
    total += float(car['cty'])

average = total / len(mpg)

print(average)
```

Find the average hwy fuel economy across all cars

```
total=0

for h in mpg:
  total+=float(h['hwy'])

average = total/len(mpg)
print(average)
```

Use set to return the unique values for the number of cylinders the cars.

```
cylinders = set()

for d in mpg:
    cylinders.add(d['cyl'])

print(cylinders)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```
MpgByCyl = []

for c in cylinders:
    summpg = 0
    typecount = 0

    for d in mpg:
        if d['cyl'] == c:
            summpg += float(d['cty'])
            typecount += 1

    MpgByCyl.append((c, summpg / typecount))

MpgByCyl.sort(key=lambda x: x[0])
MpgByCyl
```

Return the unique values for the class types in our dataset using set.

```
vehicleclass = set(d['class'] for d in mpg)
print(vehicleclass)
```

Find average miles per gallon in the highway for each class of vehicles

```
MileageByClass = []

for t in vehicleclass:
    sumpg = 0
    classCount = 0

    for d in mpg:
        if d['class'] == t:
            sumpg += float(d['hwy'])
            classCount += 1

    MileageByClass.append((t, sumpg / classCount))

MileageByClass
```
